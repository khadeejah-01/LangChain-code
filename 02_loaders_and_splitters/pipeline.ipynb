## Document Loader and Splitter in a pipeline

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter


# -----------------------------------
# 1. Load the PDF
# -----------------------------------

loader = PyPDFLoader("../data/NIPS-2017-attention-is-all-you-need-Paper.pdf")

docs = loader.load()

print("Number of documents/pages:", len(docs)) # each page of pdf is made a langchain document
#print(docs[0].page_content)
#print(docs[1].metadata)

# -----------------------------------
# 2. Initialize the text splitter
# -----------------------------------

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
)


# -----------------------------------
# 3. Split the loaded documents
# -----------------------------------

chunks = splitter.split_documents(docs)


# -----------------------------------
# 4. Inspect the chunks
# -----------------------------------

print("Number of chunks:", len(chunks))

print("\nFirst chunk:")
print(chunks[0].page_content)
```
